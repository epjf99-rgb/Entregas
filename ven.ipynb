```python
# prompt: grafica la funcion

import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

# Assuming 'marketing.csv' is in the current working directory
# If not, provide the full path to the file
try:
    data = pd.read_csv("marketing.csv")
except FileNotFoundError:
    print("Error: 'marketing.csv' not found. Please ensure the file is in the correct location.")
    exit()


# Assuming 'newspaper' and 'sales' columns exist in your DataFrame
# Replace 'sales' with the actual column name if different
plt.figure(figsize=(10, 6))
plt.scatter(data['newspaper'], data['sales'], marker='o', color='blue', label='Data Points')

# Fit the linear regression model
funcion = LinearRegression()
funcion.fit(data[['newspaper']], data['sales'])

# Generate points for the regression line
x_values = np.linspace(min(data['newspaper']), max(data['newspaper']), 100)  # 100 points for a smooth line
y_values = funcion.predict(x_values.reshape(-1, 1))

# Plot the regression line
plt.plot(x_values, y_values, color='peru', label='Regression Line')


plt.title('Newspaper Spending vs Sales')
plt.xlabel('Newspaper Spending')
plt.ylabel('Sales')
plt.grid(True)
plt.legend() # Show labels for data points and regression line
plt.show()

print(f'ordenada: {funcion.intercept_}')
print(f'pendiente: {funcion.coef_}')
print(f'Regression Equation: f(x) = {funcion.coef_[0]}x + {funcion.intercept_}')
```

```python
funcion.score(data[["newspaper"]],data["sales"])
```

```python
ax=sns.regplot(x="newspaper",y="sales",data=data,ci=95)
ax.grid()
plt.show()
```

el sombreado son las betas ubicadas en la región a lo mas que podemos aspirar segun el intervalo de confianza para que exista que tan relacionados estan los datos

```python

plt.figure(figsize=(10, 6))
plt.scatter(data['youtube'], data['sales'], marker='o', color='blue', label='Data Points')

# Fit the linear regression model
funcion = LinearRegression()
funcion.fit(data[['youtube']], data['sales'])

# Generate points for the regression line
x_values = np.linspace(min(data['youtube']), max(data['youtube']), 100)  # 100 points for a smooth line
y_values = funcion.predict(x_values.reshape(-1, 1))

# Plot the regression line
plt.plot(x_values, y_values, color='peru', label='Regression Line')

plt.title('YouTube Spending vs Sales')
plt.xlabel('YouTube Spending')
plt.ylabel('Sales')
plt.grid(True)
plt.legend()
plt.show()

print(f'ordenada: {funcion.intercept_}')
print(f'pendiente: {funcion.coef_}')
print(f'Regression Equation: f(x) = {funcion.coef_[0]}x + {funcion.intercept_}')
print(f"R-squared: {funcion.score(data[['youtube']], data['sales'])}")

ax = sns.regplot(x="youtube", y="sales", data=data, ci=95)
ax.grid()
plt.show()
```
