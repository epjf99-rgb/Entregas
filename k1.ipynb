```python
import pandas
import numpy
from matplotlib import pyplot
from sklearn.neural_network import MLPClassifier
from tensorflow.keras.datasets.mnist import load_data
```

```python
load_data() #
```

```python
(Xt,yt),(Xtest,ytest)=load_data()
print(f"Xt.shape={Xt.shape}, yt={yt.shape}")
print(f"Xtest.shape={Xtest.shape}, ytest={ytest.shape}")#hola mundo de reconocimiento de imagenes esta en escala de grises
```

```python
for i in range(9):
    pyplot.subplot(3,3,i+1)
    pyplot.imshow(Xt[i],cmap=pyplot.get_cmap('gray'))
    pyplot.title(label=f"Imagen{i}")
    pyplot.tight_layout()
pyplot.show()
```

```python
numpy.unique(yt)
```

```python
red=MLPClassifier(verbose=1 )
```

```python
X=[]
for i in Xt:
  X.append(numpy.concatenate(i))
X=numpy.array(X)
X.shape
```

```python
red.fit(X,yt)
```

```python
red.score(X,yt)
```

```python
red.score(Xtest.reshape(10000,784),ytest)
```

```python
fig=pyplot.figure(figsize=(10,10))

for i in range(25):
    e=0  # Indented this line to be inside the loop
    pyplot.subplot(5,5,i+1)
    pyplot.imshow(Xt[i+e],cmap=pyplot.get_cmap('gray'))
    pyplot.title(label=f"Real{yt[i+e]}, {red.predict(Xt[i+e].reshape(1, -1))}") # Reshape to (1, 784) for prediction
    pyplot.tight_layout()
pyplot.show() # Call pyplot.show() to display the figure
```
