```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

def graficar_multiseries(series_dict):
    fig, ax = plt.subplots(figsize=(30, 7))
    scaler = MinMaxScaler()  # Inicializar el escalador

    # Iterar sobre cada serie en el diccionario
    for label, series in series_dict.items():
        series_scaled = scaler.fit_transform(series.values.reshape(-1, 1))  # Escalar los valores
        plt.plot(series.index, series_scaled, label=label)  # Graficar cada serie

    plt.xlabel("Día")
    plt.ylabel("Valor Escalado")
    plt.title("Series Escaladas")
    plt.legend()
    plt.grid()
    plt.show()

# Ejemplo de uso (usando nombres de columnas ficticias)
f = pd.read_csv("facebook.csv")["facebook"]
y = pd.read_csv("youtube.csv")["youtube"]
periodico = pd.read_csv("periodico.csv", encoding="latin2")["periódico"]
v = pd.read_csv("ventas.csv")["ventas"]

# Llamar a la función con las series de datos
series_dict = {
    "Facebook": f,
    "YouTube": y,
    "Periódico": periodico,
    "Ventas": v
}

graficar_multiseries(series_dict)
```

```python


import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

def graficar_multiseries(series_dict):
    fig, ax = plt.subplots(figsize=(12, 6))  # Ajustar tamaño de la figura
    scaler = MinMaxScaler()

    for label, series in series_dict.items():
        series_scaled = scaler.fit_transform(series.values.reshape(-1, 1))
        ax.plot(series.index, series_scaled, label=label)

    ax.set_xlabel("Día")
    ax.set_ylabel("Valor Escalado")
    ax.set_title("Comparación de Inversiones")
    ax.legend()
    ax.grid(True)
    plt.show()

    # Calcular la rentabilidad promedio de cada serie
    rentabilidades_promedio = {}
    for label, series in series_dict.items():
      rentabilidades_promedio[label] = series.mean()

    print("Rentabilidad promedio de cada inversión:")
    for label, rentabilidad in rentabilidades_promedio.items():
        print(f"- {label}: {rentabilidad}")

    # Determinar la mejor inversión
    mejor_inversion = max(rentabilidades_promedio, key=rentabilidades_promedio.get)
    print(f"\nLa mejor inversión, según la rentabilidad promedio, es: {mejor_inversion}")



try:
    f = pd.read_csv("facebook.csv")["facebook"]
    y = pd.read_csv("youtube.csv")["youtube"]
    periodico = pd.read_csv("periodico.csv", encoding="latin2")["periódico"]
    v = pd.read_csv("ventas.csv")["ventas"]
except FileNotFoundError:
    print("Error: Uno o más archivos CSV no se encontraron. Por favor, verifica las rutas.")
    exit()


# Crear un diccionario con las series
series_dict = {
    "Facebook": f,
    "YouTube": y,
    "Periódico": periodico,
    "Ventas": v
}

# Graficar y analizar las series
graficar_multiseries(series_dict)
```
