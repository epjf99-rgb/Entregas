```python
pt_tokens = tokenizer(input_text, return_tensors="pt")
```

```python
import torch
with torch.no_grad():
    outputs = model(**pt_tokens) #El modelo va a generar embedding
    last_hidden_states = outputs.last_hidden_state #C
print("Token embeddings:") #D
for token_id, embedding in zip(pt_tokens["input_ids"][0], last_hidden_states[0]):
    word = tokenizer.decode(int(token_id))
    print(f"{word}: {embedding[:5]}") # Print only the first 5 elements of the embedding for brevity
```
