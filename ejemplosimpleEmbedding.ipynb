```python
import tensorflow as tf
from tensorflow.keras.layers import TextVectorization

words = ["determination", "undyne", "papyrus", "sans"]
vectorizer = TextVectorization(max_tokens=8)
vectorizer.adapt(words)
vectorized_words = vectorizer(words)
print(vectorized_words)
```

```python
from tensorflow.keras.layers import Embedding

embedding_layer = Embedding(input_dim=vectorizer.vocabulary_size(), output_dim=16)
embedded_words = embedding_layer(vectorized_words)
print(embedded_words)
```

```python
embed_weights = embedding_layer.get_weights()[0]
print(embed_weights)

words_in_vocab = vectorizer.get_vocabulary()
print(words_in_vocab)
```

```python
import io

out_v = io.open("embedding_vectors.tsv", "w", encoding="utf-8")
out_m = io.open("embedding_metadata.tsv", "w", encoding="utf-8")

for num, word in enumerate(words_in_vocab):
   if num == 0:
      continue  # skip padding token
   vec = embed_weights[num]
   out_m.write(word + "\n")
   out_v.write("\t".join([str(x) for x in vec]) + "\n")

out_v.close()
out_m.close()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Get the non-padding/unknown word vectors and their corresponding words
word_vectors = embed_weights[2:]  # skip padding and unknown tokens
words_in_vocab_filtered = words_in_vocab[2:]

# Calculate cosine similarity
similarity_matrix = cosine_similarity(word_vectors)

# Display the similarity matrix with labels
import pandas as pd
similarity_df = pd.DataFrame(similarity_matrix, index=words_in_vocab_filtered, columns=words_in_vocab_filtered)
```
