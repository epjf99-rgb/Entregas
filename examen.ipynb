```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt


datos = pd.read_csv("Metro_Interstate_Traffic_Volume.csv")

# Convert 'date_time' to datetime and extract hour, day of the week
datos['date_time'] = pd.to_datetime(datos['date_time'])
datos['hour'] = datos['date_time'].dt.hour
datos['day_of_week'] = datos['date_time'].dt.dayofweek
# Convert temperature from Kelvin to Celsius
datos['temp_C'] = datos['temp'] - 273.15

numerical_features = ['temp', 'rain_1h', 'clouds_all', 'hour', 'day_of_week', 'temp_C']

for feature in numerical_features:
  # Prepare data for linear regression
  X = datos[[feature]].values
  y = datos['traffic_volume'].values

  # Create and train the model
  model = LinearRegression()
  model.fit(X, y)

  # Generate predictions for the regression line
  X_range = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
  y_pred = model.predict(X_range)


  plt.figure(figsize=(10, 6))
  plt.scatter(X, y, color='lightgray', alpha=0.5, label='Datos observados')
  plt.plot(X_range, y_pred, color='red', linewidth=2, label='Línea de regresión')
  plt.xlabel(feature, fontsize=12)
  plt.ylabel('Volumen de tráfico', fontsize=12)
  plt.title(f'Regresión Lineal: Tráfico vs. {feature}', fontsize=14)
  plt.legend()
  plt.tight_layout()
  plt.show()


  # Output the slope and intercept of the model
  print(f"Coeficiente (pendiente) for {feature}: {model.coef_[0]:.2f}")
  print(f"Intercepto for {feature}: {model.intercept_:.2f}")
  print("-" * 50)
```

```python

datos = pd.read_csv("Metro_Interstate_Traffic_Volume.csv")


datos['date_time'] = pd.to_datetime(datos['date_time'])
datos['hour'] = datos['date_time'].dt.hour
datos['day_of_week'] = datos['date_time'].dt.dayofweek

datos['temp_C'] = datos['temp'] - 273.15

numerical_features = ['temp', 'rain_1h', 'clouds_all', 'hour', 'day_of_week', 'temp_C']

for feature in numerical_features:

  X = datos[[feature]].values
  y = datos['traffic_volume'].values


  model = LinearRegression()
  model.fit(X, y)


  X_range = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
  y_pred = model.predict(X_range)


  r_squared = model.score(X, y)


  plt.figure(figsize=(10, 6))
  plt.scatter(X, y, color='lightgray', alpha=0.5, label='Datos observados')
  plt.plot(X_range, y_pred, color='red', linewidth=2, label='Línea de regresión')
  plt.xlabel(feature, fontsize=12)
  plt.ylabel('Volumen de tráfico', fontsize=12)
  plt.title(f'Regresión Lineal: Tráfico vs. {feature}', fontsize=14)
  plt.legend()
  plt.grid(True)
  plt.tight_layout()
  plt.show()


  print(f"Coeficiente (pendiente) for {feature}: {model.coef_[0]:.2f}")
  print(f"Intercepto for {feature}: {model.intercept_:.2f}")
  print(f"R-squared for {feature}: {r_squared:.2f}")
  print("-" * 50)
```
