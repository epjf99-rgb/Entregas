```python
import math

class Punto3D:
    def __init__(self, x=0, y=0, z=0):
        """Constructor que inicializa las coordenadas en 3D"""
        self.x = x
        self.y = y
        self.z = z

    def copy(self):
        """Devuelve una copia del punto actual"""
        return Punto3D(self.x, self.y, self.z)

    def distancia(self, otro):
        """Calcula la distancia euclidiana entre este punto y otro"""
        return math.sqrt((self.x - otro.x) ** 2 + (self.y - otro.y) ** 2 + (self.z - otro.z) ** 2)

    def puntoMedio(self, otro):
        """Devuelve el punto medio entre este punto y otro"""
        return Punto3D((self.x + otro.x) / 2, (self.y + otro.y) / 2, (self.z + otro.z) / 2)

    def __repr__(self):
        """Representación en texto del punto"""
        return f"Punto3D(x={self.x}, y={self.y}, z={self.z})"

punto1 = Punto3D(1, 2, 3)
punto2 = Punto3D(4, 5, 6)
punto3 = Punto3D(-1, -2, -3)


punto_copiado = punto1.copy()

distancia_p1_p2 = punto1.distancia(punto2)

punto_medio_p1_p2 = punto1.puntoMedio(punto2)


print(f"Punto 1: {punto1}")
print(f"Punto 2: {punto2}")
print(f"Punto 3: {punto3}")
print(f"Copiado de Punto 1: {punto_copiado}")
print(f"Distancia entre Punto 1 y Punto 2: {distancia_p1_p2}")
print(f"Punto medio entre Punto 1 y Punto 2: {punto_medio_p1_p2}")


```

```python
import math
import matplotlib.pyplot as plt

class Punto3D:
    def __init__(self, x=0, y=0, z=0):
        """Constructor que inicializa las coordenadas en 3D"""
        self.x = x
        self.y = y
        self.z = z

    def copy(self):
        """Devuelve una copia del punto actual"""
        return Punto3D(self.x, self.y, self.z)

    def distancia(self, otro):
        """Calcula la distancia euclidiana entre este punto y otro"""
        return math.sqrt((self.x - otro.x) ** 2 + (self.y - otro.y) ** 2 + (self.z - otro.z) ** 2)

    def puntoMedio(self, otro):
        """Devuelve el punto medio entre este punto y otro"""
        return Punto3D((self.x + otro.x) / 2, (self.y + otro.y) / 2, (self.z + otro.z) / 2)

    def __repr__(self):
        """Representación en texto del punto"""
        return f"Punto3D(x={self.x}, y={self.y}, z={self.z})"


fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

punto1 = Punto3D(1, 2, 3)
punto2 = Punto3D(4, 5, 6)
punto3 = Punto3D(-1, -2, -3)


punto_copiado = punto1.copy()

distancia_p1_p2 = punto1.distancia(punto2)

punto_medio_p1_p2 = punto1.puntoMedio(punto2)


print(f"Punto 1: {punto1}")
print(f"Punto 2: {punto2}")
print(f"Punto 3: {punto3}")
print(f"Copiado de Punto 1: {punto_copiado}")
print(f"Distancia entre Punto 1 y Punto 2: {distancia_p1_p2}")
print(f"Punto medio entre Punto 1 y Punto 2: {punto_medio_p1_p2}")

ax.scatter(punto1.x, punto1.y, punto1.z, c='r', marker='o', label='Punto 1')
ax.scatter(punto2.x, punto2.y, punto2.z, c='b', marker='o', label='Punto 2')
ax.scatter(punto3.x, punto3.y, punto3.z, c='g', marker='o', label='Punto 3')
ax.scatter(punto_medio_p1_p2.x, punto_medio_p1_p2.y, punto_medio_p1_p2.z, c='black', marker='o', label='Punto Medio')



ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.title("Mi primer pyplot 3D")
ax.legend()


plt.show()
```
